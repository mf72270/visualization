```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load and prepare the data
df = pd.read_csv('diabetes_clinics.csv')
```

```python
# Count clinics by city
city_counts = df['City'].value_counts().head(15)  # Top 15 cities
# Create the visualization
plt.figure(figsize=(12, 8))
bars = plt.bar(city_counts.index, city_counts.values, color='skyblue', edgecolor='navy')
# Customize the plot
plt.title('Diabetes Education Clinics Distribution Across Ontario Cities', 
          fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Cities', fontsize=12, fontweight='bold')
plt.ylabel('Number of Clinics', fontsize=12, fontweight='bold')
plt.xticks(rotation=45, ha='right')

# Add value labels on bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + 0.1,
             f'{int(height)}', ha='center', va='bottom')
  

"""
This whole block
   # ---Add annotation for Toronto
toronto_index = list(city_counts.index).index('Toronto')
toronto_count = city_counts['Toronto']
plt.annotate('Toronto leads with\n15 diabetes clinics', 
             xy=(toronto_index, toronto_count),
             xytext=(toronto_index + 2, toronto_count + 5),
             fontsize=9,
             fontweight='bold',
             color='darkred',
             bbox=dict(boxstyle='round,pad=0.3', facecolor='yellow', alpha=0.9),
             arrowprops=dict(arrowstyle='->', connectionstyle='arc3,rad=0.3', 
                           color='darkred', lw=1))

    # ---Add annotation for Toronto---
"""
# Add annotation for Toronto
toronto_index = list(city_counts.index).index('Toronto')
toronto_count = city_counts['Toronto']
plt.annotate('Toronto leads with\n15 diabetes clinics', 
             xy=(toronto_index - 0.3, toronto_count),  # Point to left side of bar
             xytext=(toronto_index - 2.5, toronto_count + 3),  # Position text to the left
             fontsize=9,
             fontweight='bold',
             color='darkred',
             bbox=dict(boxstyle='round,pad=0.3', facecolor='yellow', alpha=0.9),
             arrowprops=dict(arrowstyle='->', connectionstyle='arc3,rad=0.3', 
                           color='darkred', lw=1))

# Improve accessibility
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()

 # Save the plot
plt.savefig('diabetes_clinics_distribution.png', dpi=300, bbox_inches='tight')
plt.show()

    
```
